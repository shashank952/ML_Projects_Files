```python
import cv2
import os
import numpy as np

# ===== USER INPUT =====
input_folder = r"C:\Users\sanja\Desktop\MRI_Project\Pleural Effusion\dataset\Alt_Org_Dataset"     # <- put your folder path
output_folder = r"C:\Users\sanja\Desktop\MRI_Project\Pleural Effusion\dataset\Croped_Org_Dataset"       # <- processed images will be saved here
TARGET_SIZE = (224, 224)                   # change if CNN needs different size
# ======================

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

for file in os.listdir(input_folder):
    if file.lower().endswith((".png", ".jpg", ".jpeg", ".bmp")):
        img_path = os.path.join(input_folder, file)
        img = cv2.imread(img_path)

        if img is None:
            print(f"Skipping {file} (cannot read)")
            continue
        
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

        # Threshold to remove black background
        _, thresh = cv2.threshold(gray, 10, 255, cv2.THRESH_BINARY)

        coords = cv2.findNonZero(thresh)
        x, y, w, h = cv2.boundingRect(coords)

        cropped = img[y:y+h, x:x+w]

        # Resize for CNN
        resized = cv2.resize(cropped, TARGET_SIZE, interpolation=cv2.INTER_AREA)

        save_path = os.path.join(output_folder, file)
        cv2.imwrite(save_path, resized)

        print(f"Processed: {file}")

print("Done. All images saved to:", output_folder)
```
