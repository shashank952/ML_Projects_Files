```python
# Cell 0: Force PyTorch to use NVIDIA GPU only
import os

# Hide AMD GPU (GPU 0) and expose only NVIDIA GPU (GPU 1)
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

print("CUDA_VISIBLE_DEVICES set to:", os.environ["CUDA_VISIBLE_DEVICES"])
```

```python
import torch
print("CUDA Available:", torch.cuda.is_available())
print("GPU Count:", torch.cuda.device_count())
print("Using GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None")
```
