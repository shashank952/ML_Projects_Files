```python
# Import necessary libraries
import pandas as pd
import numpy as np
from pathlib import Path

# Load the datasets
test_categorical = pd.read_excel("C:/Users/sanja/Desktop/ppt's and pdf/WiDS/TEST/TEST_CATEGORICAL.xlsx")
test_quant = pd.read_excel("C:/Users/sanja/Desktop/ppt's and pdf/WiDS/TEST/TEST_QUANTITATIVE_METADATA.xlsx")

# Display initial info about the datasets
print("Categorical Dataset Info:")
print(test_categorical.info())
print("\nQuantitative Dataset Info:")
print(test_quant.info())

# Function for basic data cleaning
def clean_data(df, is_categorical=False):
    # Make a copy of the dataframe
    cleaned_df = df.copy()
    
    # Handle missing values
    if is_categorical:
        # For categorical data, fill missing with mode or 'Unknown'
        for col in cleaned_df.columns:
            if cleaned_df[col].dtype == 'object':
                cleaned_df[col] = cleaned_df[col].fillna('Unknown')
            else:
                # If numeric categorical, fill with mode
                cleaned_df[col] = cleaned_df[col].fillna(cleaned_df[col].mode()[0])
    else:
        # For quantitative data, fill with mean or median (depending on distribution)
        for col in cleaned_df.select_dtypes(include=np.number).columns:
            # Check skewness to decide between mean or median
            skewness = cleaned_df[col].skew()
            if abs(skewness) > 1:  # Highly skewed
                cleaned_df[col] = cleaned_df[col].fillna(cleaned_df[col].median())
            else:
                cleaned_df[col] = cleaned_df[col].fillna(cleaned_df[col].mean())
    
    # Remove duplicate rows
    cleaned_df = cleaned_df.drop_duplicates()
    
    # Standardize column names (remove special characters, make lowercase, etc.)
    cleaned_df.columns = cleaned_df.columns.str.lower().str.replace('[^a-z0-9_]', '_', regex=True)
    
    return cleaned_df

# Clean both datasets
cleaned_categorical = clean_data(test_categorical, is_categorical=True)
cleaned_quant = clean_data(test_quant, is_categorical=False)

# Display cleaned data info
print("\nCleaned Categorical Dataset Info:")
print(cleaned_categorical.info())
print("\nCleaned Quantitative Dataset Info:")
print(cleaned_quant.info())

# Save cleaned datasets to new CSV files
output_dir = "C:/Users/sanja/Desktop/Test_output"
Path(output_dir).mkdir(parents=True, exist_ok=True)  # Create directory if it doesn't exist

categorical_path = output_dir + "cleaned_categorical.csv"
quant_path = output_dir + "cleaned_quantitative.csv"

cleaned_categorical.to_csv(categorical_path, index=False)
cleaned_quant.to_csv(quant_path, index=False)

# Print the paths where the cleaned files are saved
print("\nCleaned files saved at:")
print(f"Categorical data: {categorical_path}")
print(f"Quantitative data: {quant_path}")
```

```python
import pandas as pd
import os

# File paths (update these with your actual paths)
test_cat_path = r"C:/Users/sanja/Desktop/ppt's and pdf/WiDS/TEST/TEST_CATEGORICAL.xlsx"
test_quant_path = r"C:/Users/sanja/Desktop/ppt's and pdf/WiDS/TEST/TEST_QUANTITATIVE_METADATA.xlsx"
test_func_path = r"C:/Users/sanja/Desktop/ppt's and pdf/WiDS/TEST/TEST_FUNCTIONAL_CONNECTOME_MATRICES.csv"

# Output path for merged file
output_dir = r"C:/Users/sanja/Desktop/Test_output"
output_path = os.path.join(output_dir, "test_merged.csv")

# Create directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Read Excel + CSV correctly
print("Loading files...")
test_cat = pd.read_excel(test_cat_path)   # Excel
test_quant = pd.read_excel(test_quant_path)  # Excel
test_func = pd.read_csv(test_func_path)   # CSV

# Verify shapes before merging
print("\nOriginal shapes:")
print(f"Categorical data: {test_cat.shape}")
print(f"Quantitative data: {test_quant.shape}")
print(f"Functional data: {test_func.shape}")

# Merge categorical and quantitative data first
print("\nMerging categorical and quantitative data...")
merged = pd.merge(test_cat, test_quant, 
                 on="participant_id", 
                 how="inner",
                 validate="one_to_one")

# Then merge with functional data
print("Merging with functional data...")
final_merged = pd.merge(merged, test_func,
                       on="participant_id",
                       how="inner",
                       validate="one_to_one")

# Save merged file
final_merged.to_csv(output_path, index=False)

print(f"\n✅ Successfully merged files! Final shape: {final_merged.shape}")
print(f"Merged file saved to: {output_path}")

# Show first few participant IDs as verification
print("\nFirst few participant IDs in merged file:")
print(final_merged["participant_id"].head())
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.multioutput import MultiOutputClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

def load_data():
    train_data = pd.read_csv(r"C:\Users\sanja\Desktop\Train_Output\test_merged.csv")
    target_data = pd.read_excel(r"C:\Users\sanja\Desktop\ppt's and pdf\WiDS\TRAIN_NEW\TRAINING_SOLUTIONS.xlsx")
    test_data = pd.read_csv(r"C:\Users\sanja\Desktop\Test_Output\test_merged.csv")
    
    train = pd.merge(train_data, target_data, on="participant_id")
    return train, test_data

# --- Preprocess data ---
def preprocess(train, test_data):
    # Training data
    X_train = train.drop(columns=["participant_id", "ADHD_Outcome", "Sex_F"])
    y_train = train[["ADHD_Outcome", "Sex_F"]]
    
    # Test data
    X_test = test_data.drop(columns=["participant_id"])
    participant_ids = test_data["participant_id"]
    
    # --- Handle missing values ---
    imputer = SimpleImputer(strategy="median")
    X_train_imputed = imputer.fit_transform(X_train)
    X_test_imputed = imputer.transform(X_test)
    
    # --- Scale features ---
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_imputed)
    X_test_scaled = scaler.transform(X_test_imputed)
    
    return X_train_scaled, y_train, X_test_scaled, participant_ids

# --- Train model ---
def train_model(X_train, y_train):
    model = MultiOutputClassifier(
        LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    )
    model.fit(X_train, y_train)
    return model

# --- Make predictions ---
def predict(model, X_test, participant_ids):
    y_pred = model.predict(X_test)
    submission = pd.DataFrame(y_pred, columns=["ADHD_Outcome", "Sex_F"])
    submission.insert(0, "participant_id", participant_ids)
    return submission

# --- Main execution ---
if __name__ == "__main__":
    # Load data
    train, test_data = load_data()
    
    # Preprocess
    X_train, y_train, X_test, participant_ids = preprocess(train, test_data)
    
    # Train
    model = train_model(X_train, y_train)
    
    # Predict
    submission = predict(model, X_test, participant_ids)
    
    # Validate and save
    if submission.shape[0] != 304:
        print(f"Error: Expected 304 rows, got {submission.shape[0]}")
    else:
        submission_path = r"C:/Users/sanja/Desktop/SUBS/Sub1.csv"
        submission.to_csv(submission_path, index=False)
        print(f"Submission saved to {submission_path}")
        print("\nSubmission preview:")
        print(submission.head())
```
